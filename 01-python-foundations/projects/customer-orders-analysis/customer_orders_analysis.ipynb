**Task 1: Store customer orders**

• Create a list of customer names

```python
customer_names = []

num_cust = int(input("How many customers do you have? "))


for i in range(num_cust):
  name = input(f"Enter customer {i + 1} name: ")
  customer_names.append(name)
```

```python
print("Customer Names: ",customer_names)
```

• Store each customer's order details (customer name, product, price, category) as tuples inside a list

```python
orders = [
  (customer_names[0], "Laptop", 999.99, "Electronics"),
  (customer_names[1], "T-shirt", 79.99, "Clothing"),
  (customer_names[2], "Coffee Maker", 49.99, "Kitchen"),
  (customer_names[3], "Running Shoes", 109.99, "Clothing"),
  (customer_names[4], "Electric Drill", 240.00, "Hardware"),
  (customer_names[5], "Bath Towel", 19.99, "Home Goods")
]
orders
```

• Use a dictionary where keys are customer names and values are lists of ordered
products

```python
customer_orders = {}

for order in orders:
  customer = order[0]
  product = order[1]

  if customer not in customer_orders:
    customer_orders[customer] = []

  customer_orders[customer].append(product)

print(customer_orders)
```


**Task 2: Classify products by category**

• Use a dictionary to map each product to its respective category


```python
product_categories = {}

for order in orders:
  customer, product, price, category = order
  product_categories[product] = category

product_categories
```

• Create a set of unique product categories


```python
unique_categories = set(product_categories.values())
unique_categories
```


• Display all available product categories

```python
print("\nAvailable Product Categories: ", unique_categories)
```

**Task 3: Analyze customer orders**

• Use a loop to calculate the total amount each customer spends


```python
customer_spending = {}

for order in orders:
  customer, product, price, category = order

  if customer in customer_spending:
    customer_spending[customer] += price # Taking care of cases where persons have >1 orders
  else:
    customer_spending[customer] = price

customer_spending
```

• If the total purchase value is above $100, classify the customer as a high-value buyer

• If it is between 50 and $100, classify the customer as a moderate buyer

• If it is below $50, classify them as a low-value buyer


```python
for customer, total in customer_spending.items():

  if total > 100:
    classification = "High-Value Buyer"

  elif total >= 50:
    classification = "Moderate-Value Buyer"

  else:
    classification = "Low-Value Buyer"

  print(f"{customer}: ${total:.2f} - {classification}")
```

**Task 4. Generate Business Insights**

• Calculate the total revenue per product category and store it in a dictionary


```python
category_revenue = {}

for order in orders:
  customer, product, price, category = order

  if category in category_revenue:
    category_revenue[category] += price
  else:
    category_revenue[category] = price

category_revenue
```


• Extract unique products from all orders using a set



```python
unique_products = set(order[1] for order in orders)
unique_products
```

• Use a list comprehension to find all customers who purchased electronics



```python
electronics_customers = [
  customer
  for customer, product, price, category in orders
  if category == "Electronics"
]
electronics_customers
```

• Identify the top three highest-spending customers using sorting

```python
top_customers = sorted(
    customer_spending.items(),
    key=lambda x: x[1],
    reverse=True
)
top_customers[:3]
```

**Task 5. Organize and display data**

• Print a summary of each customer’s total spending and their classification



```python

for customer, total in customer_spending.items():

  if total > 100:
    classification = "High-Value Buyer"

  elif total >= 50:
    classification = "Moderate-Value Buyer"

  else:
    classification = "Low-Value Buyer"

  print(f"{customer}: ${total:.2f} - {classification}")
```

• Use set operations to find customers who purchased from multiple categories



```python
customer_categories = {}

for customer, product, price, category in orders:
  if customer not in customer_categories:
    customer_categories[customer] = set()

  customer_categories[customer].add(category)


multiple_category_customers = {
  customer
  for customer, categories in customer_categories.items()
  if len(categories) > 1
}

print("\nCustomers Who Purchased From Multiple Categories:")
print(multiple_category_customers)
```

• Identify common customers who bought both electronics and clothing

```python
electronics_customers = {
    customer for customer, product, price, category in orders
  if category == "Electronics"
}

clothing_customers = {
  customer for customer, product, price, category in orders
  if category == "Clothing"
}

common_customers = electronics_customers & clothing_customers

print("\nCustomers Who Bought Both Electronics and Clothing:")
print(common_customers)
```
